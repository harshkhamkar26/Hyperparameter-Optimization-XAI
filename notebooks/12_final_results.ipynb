# Stage 12 — Final Results and Research Synthesis

## 1. Objective

Stage 12 consolidates the experimental results produced throughout the project into a final research-oriented results structure.

The purpose is to connect:

into one reproducible experimental narrative.

## 2. Research Problem

The project investigates an explainable hyperparameter optimization framework for imbalanced tabular classification using the UCI Bank Marketing dataset.

The current case study focuses on predicting whether a customer subscribes to a term deposit.

The target variable is:

`y`

with:

- `no` = negative class
- `yes` = positive class

---

## 3. Dataset

The dataset used throughout the project is:

`UCI Bank Marketing — bank-full.csv`

The dataset contains 45,211 observations and 17 columns.

The positive class is substantially smaller than the negative class, motivating the use of imbalance-aware evaluation.

---

## 4. Experimental Pipeline

The research pipeline consists of the following stages:

### Stage 1 — Data Understanding

The dataset structure, target variable, feature types, missing-value representation, and class distribution were examined.

### Stage 2 — Exploratory Data Analysis

The distributions and relationships between variables were investigated.

The analysis was descriptive and was not interpreted as causal evidence.

### Stage 3 — Preprocessing

Numerical variables were standardized and categorical variables were one-hot encoded.

The preprocessing workflow was designed to prevent information leakage.

### Stage 4 — Baseline Models

Initial Logistic Regression, Decision Tree, and Random Forest models were evaluated without hyperparameter optimization.

### Stage 5 — Imbalance Strategies

Class weighting and SMOTENC-based approaches were evaluated to investigate strategies for handling class imbalance.

Resampling was restricted to training data to prevent leakage into the test set.

### Stage 6 — Hyperparameter Optimization

Optuna was used to optimize model hyperparameters.

F1 was used as the primary optimization objective.

### Stage 7 — Multi-Objective HPO

F1 and PR-AUC were optimized simultaneously.

This stage investigates the trade-off between minority-class classification performance and positive-class ranking performance.

The resulting Pareto solutions should not be interpreted as having a universal single optimum.

### Stage 8 — Threshold Optimization

Out-of-fold training probabilities were used to select a threshold that maximized F1.

The selected threshold was subsequently evaluated on the untouched test set.

### Stage 9 — Robustness Analysis

Model performance was evaluated using five random seeds and three stratified folds per seed.

Mean and standard deviation were calculated for the main evaluation metrics.

### Stage 10 — SHAP Explainability

SHAP was used to analyze both global and local model behavior.

Global analysis identifies features with large average attribution magnitude.

Local analysis explains individual predictions.

### Stage 11 — Explanation Stability

SHAP feature importance was evaluated across 15 seed/fold experiments.

Top-10 feature overlap and Spearman rank correlation were used to study explanation consistency.

### Stage 12 — Final Synthesis

The results of all experimental components are consolidated for final interpretation.

---

## 5. Evaluation Metrics

The project evaluates multiple complementary metrics.

### Accuracy

Overall proportion of correctly classified observations.

### Precision

Proportion of predicted positive observations that are actually positive.

### Recall

Proportion of actual positive observations correctly identified.

### F1

Harmonic mean of precision and recall.

F1 is important when both false positives and false negatives matter and the positive class is underrepresented.

### Balanced Accuracy

Average recall across the two classes.

### PR-AUC

Area under the precision-recall curve.

This metric is particularly informative for imbalanced binary classification because it focuses on precision-recall behavior for the positive class.

### ROC-AUC

Measures discrimination across classification thresholds.

---

## 6. HPO Analysis

The HPO experiment investigates whether systematic hyperparameter optimization changes model performance relative to default configurations.

The results should be interpreted by comparing the actual experimental metrics rather than assuming that optimization necessarily improves every metric.

The HPO trial budgets used in the experiments should be reported exactly as executed.

In particular, the Random Forest Stage 6 experiment may contain a different number of completed trials from the originally planned budget because the Optuna study was continued.

The final paper must report the actual number of completed trials.

---

## 7. Multi-Objective Optimization

The multi-objective experiment simultaneously considers:

- F1
- PR-AUC

These objectives can produce different preferred configurations.

A configuration may improve one metric while reducing another.

Therefore, the Pareto front is interpreted as a set of trade-off solutions rather than a single universally optimal configuration.

---

## 8. Threshold Optimization

Threshold optimization separates probability estimation from the final classification decision.

The threshold is selected using out-of-fold predictions generated from the training data.

The test set is not used to select the threshold.

The selected threshold is then applied to predictions from a final model trained on the complete training set.

This provides a leakage-aware evaluation of threshold selection.

---

## 9. Robustness Analysis

Robustness analysis evaluates model performance under different random stratified validation configurations.

Five random seeds are used:

- 42
- 7
- 21
- 52
- 100

Each seed uses three stratified folds.

Therefore:

`5 × 3 = 15`

fold-level evaluations are performed.

Mean and standard deviation are used to summarize the resulting performance.

The variability measures should be considered alongside the mean performance.

---

## 10. Explainability

SHAP provides model-level explanations.

The global analysis uses mean absolute SHAP values to identify features with large average attribution magnitude.

The local analysis identifies the features that contribute to an individual prediction.

SHAP values describe model behavior.

They should not be interpreted as evidence that a feature causally causes a customer to subscribe.

---

## 11. Explanation Stability

Explanation stability investigates whether the feature attribution patterns remain similar across different validation configurations.

Three complementary approaches are used:

### Top-10 overlap

Measures how many important features are shared between experiments.

### Jaccard similarity

Measures similarity between top-10 feature sets.

### Spearman rank correlation

Measures similarity between feature-importance rankings.

These measurements provide complementary evidence about explanation consistency.

## 15. Important Limitations

### Dataset limitation

The framework is currently demonstrated using one marketing dataset.

Additional datasets would provide stronger evidence of generalization.

### Domain limitation

Results from the Bank Marketing dataset should not automatically be generalized to every marketing classification problem.

### Model limitation

SHAP explanations describe model behavior rather than causal relationships.

### Validation limitation

Robustness across the tested seeds and folds does not prove universal robustness.

### Deployment limitation

Some Bank Marketing variables, such as `duration`, may only become available during or after a customer interaction.

Therefore, deployment use cases must define the prediction timing carefully.

### Threshold limitation

The optimized threshold is selected for the stated objective, F1.

A different business objective could require a different threshold-selection criterion.

### Novelty limitation

No claim of methodological novelty should be made solely from combining established techniques.

The final novelty claim must be supported by a literature-gap analysis.
